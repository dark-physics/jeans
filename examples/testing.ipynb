```python
import jeans

import numpy as np
import matplotlib.pyplot as plt
import time as t
from mge_potential import BaryonicModel
```

Generate the baryon potential for the Jeans model

```python
bar = BaryonicModel('NGC6674', qdisk=0.1, ngauss=8)
phi = bar.potential_function(Upsilond=1.0, Upsilonb=1.0, D=None)
```

# Old Version

```python
profile = jeans.squashed(40, 10**12.6, 6, q0=1.0, Phi_b=phi, verbose=True)
```

```python
r_vals = np.logspace(-1, 2, 100)

vc = profile.V(r_vals, Lmax=0)
q = profile.q(r_vals)

density = profile.rho_sph_avg(r_vals)

fig, ax = plt.subplots(1, 3, figsize=(15, 5))
ax[0].plot(r_vals, vc)
ax[0].set_xlabel('r [kpc]')
ax[0].set_ylabel('V_c [km/s]')
ax[1].semilogx(r_vals, q)
ax[1].set_xlabel('r [kpc]')
ax[1].set_ylabel('q')
ax[2].loglog(r_vals, density)
ax[2].set_xlabel('r [kpc]')
ax[2].set_ylabel(r'$\rho$ [M$_\odot$/kpc$^3$]')
plt.tight_layout()
plt.show()
```
